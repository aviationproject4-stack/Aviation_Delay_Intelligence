### Export Gold, Silver & Bronze Tables to Parquet

Writes each Delta table out to a Parquet export volume so the data can be consumed outside Databricks (e.g. Tableau Public).

#### Bronze Layer Exports

```python
spark.table("workspace.aviation_bronze.airline_dim_raw").write.mode(
    "overwrite"
).parquet("/Volumes/workspace/aviation_bronze/aviation_exports_bronze/airline_dim_raw")
```

```python
spark.table("workspace.aviation_bronze.airport_dim_raw").write.mode(
    "overwrite"
).parquet("/Volumes/workspace/aviation_bronze/aviation_exports_bronze/airport_dim_raw")
```

```python
spark.table("workspace.aviation_bronze.delay_fact_raw").write.mode("overwrite").parquet(
    "/Volumes/workspace/aviation_bronze/aviation_exports_bronze/delay_fact_raw"
)
```
